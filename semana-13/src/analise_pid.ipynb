# Semana 13 — Análise da malha PID de luminosidade
Cole a saída CSV do monitor serial em `dados.csv` (colunas `t_ms,ref,y,u`) e execute as células.
O notebook calcula **sobressinal**, **tempo de acomodação (2 %)** e **erro em regime**, e plota resposta e esforço de controle.

📖 Conceitos: teoria da semana 13 (PID discreto, Exemplos 13.3/13.4). Amostragem: Molloy, cap. 9, Figura 9-4, p. 373.

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('dados.csv')
df['t'] = df.t_ms/1000
df.head()
```

```python
fig, ax = plt.subplots(2, 1, figsize=(9,6), sharex=True)
ax[0].plot(df.t, df.y, label='y (LDR)')
ax[0].plot(df.t, df.ref, '--', label='referência')
ax[0].set_ylabel('contagens ADC'); ax[0].legend(); ax[0].grid(True)
ax[1].plot(df.t, df.u, color='tab:red', label='u (duty)')
ax[1].set_xlabel('t (s)'); ax[1].set_ylabel('esforço u'); ax[1].legend(); ax[1].grid(True)
plt.tight_layout()
```

```python
# Métricas do degrau: considera do início até o fim do registro
ref = df.ref.iloc[-1]
y = df.y.to_numpy(); t = df.t.to_numpy()
sobressinal = 100*(y.max()-ref)/ref
faixa = 0.02*ref
fora = abs(y-ref) > faixa
t_acom = t[fora][-1] if fora.any() else t[0]
erro_regime = ref - y[-50:].mean()
print(f'Sobressinal: {sobressinal:.1f} %')
print(f'Tempo de acomodação (2 %): {t_acom:.2f} s')
print(f'Erro em regime (média das 50 últimas amostras): {erro_regime:.1f} contagens')
```
